### Activity 3: Debug the broken classifier

Here is some code that looks fine but misbehaves. Do not fix it by guessing. Trace the logic first.

Run the cell below. It should give "Good Standing" for a GPA of 3.2. Does it?

````python
student_gpa = 3.2

if student_gpa >= 2.0:
    status = "Probation"
elif student_gpa >= 3.0:
    status = "Good Standing"
elif student_gpa >= 3.5:
    status = "Dean's List"
else:
    status = "Failing"

print(f"Status: {status}")
````

**Trace it, then fix it.**

- When `student_gpa = 3.2`, is `3.2 >= 2.0` true? What happens the moment a condition is true?
- Why does the order of the conditions matter here?

> **Your analysis** (double-click to edit):

```
The problem is: A GPA of 3.2 is incorrectly classified as "Probation".

It happens because: 3.2 >= 2.0 is true, so Python runs that block and skips every remaining elif condition. The conditions must be checked from the highest GPA requirement to the lowest.
```

Now write your corrected version below.

````python
# Your fixed version here
student_gpa = 3.2

if student_gpa >= 3.5:
    status = "Dean's List"
elif student_gpa >= 3.0:
    status = "Good Standing"
elif student_gpa >= 2.0:
    status = "Probation"
else:
    status = "Failing"

print(f"Status: {status}")
````

**A better way to ask AI for debugging help.**

Instead of "what is wrong with this code?", give it your intent:

> "This code should give 'Good Standing' for a GPA of 3.2, but it returns 'Probation'. Can you help me understand why the logic is not behaving the way I expect?"

That shows you have already thought about the problem, and you will get a far more useful answer.

### Activity 5: Build an admissions decision system

Universities weigh several factors to make admission decisions. Design and build a fair, logical version.

**Design first:**

1. Choose 3 to 4 factors (GPA, test scores, activities, essay, special circumstances)
2. Decide how they combine (all required? at least 2 of 3?)
3. Set specific thresholds
4. Think about edge cases and tie-breakers

> **Your criteria** (double-click to edit):

```
Factor 1: GPA of 3.0 or higher out of 4.0

Factor 2: Admission test score of 70 or higher out of 100

Factor 3: Essay score of 7 or higher out of 10

Factor 4: Activities and responsibilities score of 6 or higher out of 10

Decision logic:
- If at least 3 of the 4 criteria are met, admit the student.
- If exactly 2 criteria are met, send the application for manual review.
- If fewer than 2 criteria are met, do not admit the student.
```

````python
# Your admissions decision system
# Tip: for "at least 2 of 3 conditions true", try counting how many are True.

# Your admissions decision system

gpa = float(input("Enter GPA (0.0–4.0): "))
test_score = float(input("Enter admission test score (0–100): "))
essay_score = float(input("Enter essay score (0–10): "))
activities_score = float(input("Enter activities score (0–10): "))
special_circumstances = input(
    "Are there documented special circumstances? (yes/no): "
).lower()

# Each comparison produces either True or False
criteria = [
    gpa >= 3.0,
    test_score >= 70,
    essay_score >= 7,
    activities_score >= 6
]

# Python counts True as 1 and False as 0
criteria_met = sum(criteria)

if criteria_met >= 3:
    decision = "Admitted"
elif criteria_met == 2 or special_circumstances == "yes":
    decision = "Manual Review"
else:
    decision = "Not Admitted"

print(f"Criteria met: {criteria_met} out of 4")
print(f"Decision: {decision}")
````

### Activity 6: Debug the tricky one

A student wrote this scholarship checker. The rule they wanted:

> Eligible if GPA is 3.0 or higher, AND (they have 50+ community service hours OR financial need), AND the essay scored 7 or higher.

Run it and test it. It does not behave the way they intended.

````python
print("=== Scholarship Eligibility Checker ===")

gpa = float(input("Enter GPA (0.0-4.0): "))
community_hours = int(input("Community service hours: "))
financial_need = input("Have financial need? (yes/no): ").lower()
essay_score = int(input("Essay score (1-10): "))

# Intended rule:
# GPA 3.0+ AND (community hours 50+ OR financial need) AND essay 7+

if gpa >= 3.0 and community_hours >= 50 or financial_need == "yes" and essay_score >= 7:
    print("Eligible for scholarship!")
else:
    print("Not eligible")
````

**Work through it.**

- Write the intended rule in plain English (you have it above).
- Python evaluates `and` before `or`. So how does the computer actually group that condition? Add brackets to show the grouping Python is really using.
- Where does that grouping differ from what the student meant?

> **Your analysis** (double-click to edit):

```
What Python actually checks:

(GPA is 3.0+ AND community hours are 50+)
OR
(financial need is "yes" AND essay score is 7+)

Why that is wrong:

The first group does not check the essay score, while the second group does not check the GPA. This could make an unqualified student eligible.

The fix:

GPA is 3.0+
AND
(community hours are 50+ OR financial need is "yes")
AND
essay score is 7+
```

Write your corrected version below.

````python
# Your corrected scholarship checker here
print("=== Scholarship Eligibility Checker ===")

gpa = float(input("Enter GPA (0.0–4.0): "))
community_hours = int(input("Community service hours: "))
financial_need = input("Have financial need? (yes/no): ").strip().lower()
essay_score = int(input("Essay score (1–10): "))

if (
    gpa >= 3.0
    and (community_hours >= 50 or financial_need == "yes")
    and essay_score >= 7
):
    print("Eligible for scholarship!")
else:
    print("Not eligible")
````

**Asking AI well here:**

> "This should check (A AND (B OR C) AND D) but it is not working. Can you explain how operator precedence is grouping my condition, and how to fix it?"

Follow up with: "Can you show me the same logic written with explicit brackets?"

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
- Give it your intent when you ask for help, especially when debugging.
