## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
(Create a complete, beginner-friendly Python 3 console program called “Smart Purchase Advisor.” The program is for a student who wants to decide whether they can afford a non-essential purchase without missing essential expenses or their monthly savings goal.

Ask the user for:

* Their name
* Monthly take-home income
* Total essential monthly expenses
* Planned monthly savings
* Price of the proposed purchase

Validate every numerical input. Income and purchase price must be greater than zero. Expenses and planned savings can be zero but cannot be negative. If the user enters invalid text or an invalid number, explain the problem and ask again instead of crashing. If the name is blank, use “Student.”

Calculate and display:

* Money available before the purchase: income minus essential expenses and planned savings
* Money remaining after the purchase
* Purchase price as a percentage of income
* Remaining money as a percentage of income

Use the following decision rules in this exact order:

1. If essential expenses plus planned savings are greater than monthly income, recommend “DO NOT BUY.”
2. Otherwise, if the purchase costs more than the available money, recommend “DO NOT BUY.”
3. Otherwise, if the purchase is more than 30% of monthly income OR it would leave less than 10% of income remaining, recommend “WAIT AND SAVE.”
4. Otherwise, if the purchase is 10% or less of monthly income AND at least 20% of income would remain, recommend “AFFORDABLE.”
5. Otherwise, recommend “CONSIDER CAREFULLY.”

The exact boundaries must be included. A purchase equal to 10% can be affordable, while one equal to 30% should not be placed in “WAIT AND SAVE” unless it leaves less than 10% of income remaining.

Print a clear budget summary with all money formatted to two decimal places and percentages formatted to one decimal place. Finish with the decision and a short explanation of why it was made.

Use functions for input validation, meaningful variable names, comments explaining the main logic, and no external libraries. Return only one complete Python code block that can be copied and run without changes. Do not ask follow-up questions.

The program should produce different recommendations for different financial situations. For example:

* Income 3000, expenses 1500, savings 300 and purchase 200 should produce “AFFORDABLE.”
* Income 3000, expenses 1500, savings 300 and purchase 500 should produce “CONSIDER CAREFULLY.”
* Income 3000, expenses 1500, savings 300 and purchase 1000 should produce “WAIT AND SAVE.”
* Income 2000, expenses 1800, savings 300 and purchase 100 should produce “DO NOT BUY.”
)
```

````python
# Paste the code your prompt produced here, then run it to show it works.

print("=== Smart Purchase Advisor ===")


def get_positive_number(message):
    """Ask repeatedly until the user enters a number greater than zero."""
    while True:
        try:
            value = float(input(message))

            if value > 0:
                return value

            print("Please enter a number greater than zero.")
        except ValueError:
            print("Invalid input. Please enter a numerical value.")


def get_non_negative_number(message):
    """Ask repeatedly until the user enters zero or a positive number."""
    while True:
        try:
            value = float(input(message))

            if value >= 0:
                return value

            print("Please enter zero or a positive number.")
        except ValueError:
            print("Invalid input. Please enter a numerical value.")


name = input("Enter your name: ").strip()

if name == "":
    name = "Student"

monthly_income = get_positive_number("Enter your monthly take-home income: $")
essential_expenses = get_non_negative_number(
    "Enter your total essential monthly expenses: $"
)
planned_savings = get_non_negative_number(
    "Enter your planned monthly savings: $"
)
purchase_price = get_positive_number(
    "Enter the price of the proposed purchase: $"
)

committed_money = essential_expenses + planned_savings
available_before_purchase = monthly_income - committed_money
remaining_after_purchase = available_before_purchase - purchase_price

purchase_percentage = (purchase_price / monthly_income) * 100
remaining_percentage = (remaining_after_purchase / monthly_income) * 100

# Apply the decision rules in order.
if committed_money > monthly_income:
    decision = "DO NOT BUY"
    reason = (
        "Your essential expenses and savings goal already exceed "
        "your monthly income."
    )
elif purchase_price > available_before_purchase:
    decision = "DO NOT BUY"
    reason = "The purchase costs more than the money you currently have available."
elif purchase_percentage > 30 or remaining_percentage < 10:
    decision = "WAIT AND SAVE"
    reason = (
        "The purchase is a large share of your income or would leave "
        "you with less than a 10% buffer."
    )
elif purchase_percentage <= 10 and remaining_percentage >= 20:
    decision = "AFFORDABLE"
    reason = (
        "The purchase uses no more than 10% of your income and leaves "
        "at least 20% remaining."
    )
else:
    decision = "CONSIDER CAREFULLY"
    reason = (
        "The purchase fits your budget, but it would still use a "
        "noticeable amount of your available money."
    )

print(f"\n=== {name}'s Budget Summary ===")
print(f"Monthly income:             ${monthly_income:,.2f}")
print(f"Essential expenses:         ${essential_expenses:,.2f}")
print(f"Planned savings:            ${planned_savings:,.2f}")
print(f"Available before purchase:  ${available_before_purchase:,.2f}")
print(f"Purchase price:             ${purchase_price:,.2f}")
print(f"Remaining after purchase:   ${remaining_after_purchase:,.2f}")
print(f"Purchase percentage:         {purchase_percentage:.1f}%")
print(f"Remaining percentage:        {remaining_percentage:.1f}%")
print(f"\nDecision: {decision}")
print(f"Reason: {reason}")
````
