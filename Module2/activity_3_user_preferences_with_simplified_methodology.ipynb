## Step 5: Write the Code

**What to do:** Now turn your pseudocode into real Python, using only `input()`, `print()`, variables, and basic operations. Below is one example solution. Run it, then compare it line by line with the pseudocode you wrote.

````python
# Step 5: Final Implementation of User Preferences program

# Get the first movie and rating
movie1 = input("Enter the first movie name: ")
rating1 = input("Enter the first movie rating: ")

# Get the second movie and rating
movie2 = input("Enter the second movie name: ")
rating2 = input("Enter the second movie rating: ")

# Build the tier list message
tier_list = f"""Movie Tier List:
1. {movie1} - {rating1}
2. {movie2} - {rating2}"""

# Print the tier list
print(tier_list)
````
