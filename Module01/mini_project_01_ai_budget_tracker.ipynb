```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:
write a budget tracker program in python that:

Takes the user's monthly income
Takes up to 3 expenses
Calculates what's left over
Warns the user if the budget goes negative
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE
income = float(input("Enter your monthly income: $"))

expense1 = float(input("Enter expense 1: $"))
expense2 = float(input("Enter expense 2: $"))
expense3 = float(input("Enter expense 3: $"))

total_expenses = expense1 + expense2 + expense3
left_over = income - total_expenses

print("\n--- Monthly Budget ---")
print(f"Income: ${income:.2f}")
print(f"Total expenses: ${total_expenses:.2f}")
print(f"Money left over: ${left_over:.2f}")

if left_over < 0:
    print("Warning: Your budget has gone negative!")
else:
    print("Your budget is within your monthly income.")
```

## Part 4: Enhancement Challenge

Pick ONE feature to add with AI's help:
- 🏷️ Expense categories (food, transport, entertainment)
- 📊 Percentage of income spent
- ✅ Input validation (no negative numbers)
- 📝 A detailed budget summary with advice


```python
# Document your enhancement choice and the prompt you used
enhancement_choice = "I chose to add:  Percentage of income spent"
enhancement_prompt = """
Enhance this code by adding a  Percentage of income spent feature in the code, that displays how much of the users income is being used up on expenses, if a person goes negative then it should show more than a 100%
"""

print(f"🚀 Enhancement: {enhancement_choice}")
```

```python
# ENHANCED BUDGET TRACKER
income = float(input("Enter your monthly income: $"))

expense1 = float(input("Enter expense 1: $"))
expense2 = float(input("Enter expense 2: $"))
expense3 = float(input("Enter expense 3: $"))

# Calculate total expenses
total_expenses = expense1 + expense2 + expense3

# Calculate money left over
left_over = income - total_expenses

# Calculate percentage of income spent
if income > 0:
    percentage_spent = (total_expenses / income) * 100
else:
    percentage_spent = 0

# Display budget summary
print("\n--- Monthly Budget ---")
print(f"Income: ${income:.2f}")
print(f"Total expenses: ${total_expenses:.2f}")
print(f"Money left over: ${left_over:.2f}")
print(f"Percentage of income spent: {percentage_spent:.2f}%")

# Check if budget is negative
if left_over < 0:
    print("Warning: Your budget has gone negative!")
else:
    print("Your budget is within your monthly income.")
```
